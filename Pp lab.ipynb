## **📌 Week 2: Basic Math & Input/Output**  

### **1️⃣ Compound Interest Calculator**  
**Formula:**  
\[ A = P \times (1 + \frac{r}{100})^n \]  

```python
P = float(input("Enter Principal Amount: "))  
r = float(input("Enter Interest Rate (%): "))  
n = float(input("Enter Time (Years): "))  
amount = P * (1 + r/100) ** n  
print("Total Amount:", amount)  
```

### **2️⃣ Distance Between Two Points**  
**Formula:**  
\[ \text{Distance} = \sqrt{(x2-x1)^2 + (y2-y1)^2} \]  

```python
import math  
x1, y1 = map(float, input("Enter Point 1 (x1 y1): ").split())  
x2, y2 = map(float, input("Enter Point 2 (x2 y2): ").split())  
distance = math.sqrt((x2-x1)**2 + (y2-y1)**2)  
print("Distance:", distance)  
```

## **📌 Week 3: Conditional Statements**  

### **1️⃣ Even or Odd Check**  
```python
num = int(input("Enter a Number: "))  
if num % 2 == 0:  
    print(num, "is Even")  
else:  
    print(num, "is Odd")  
```

### **2️⃣ Finding the Largest of 3 Numbers**  
```python
a = float(input("Enter 1st Number: "))  
b = float(input("Enter 2nd Number: "))  
c = float(input("Enter 3rd Number: "))  

if a >= b and a >= c:  
    print(a, "is Largest")  
elif b >= a and b >= c:  
    print(b, "is Largest")  
else:  
    print(c, "is Largest")  
```

## **📌 Week 4: Loops**  

### **1️⃣ Fibonacci Series (Using While Loop)**  
```python
n = int(input("How many terms? "))  
a, b = 0, 1  
count = 0  

print("Fibonacci Series:")  
while count < n:  
    print(a, end=" ")  
    next_num = a + b  
    a = b  
    b = next_num  
    count += 1  
```

### **2️⃣ Number Triangle (Using For Loop)**  
```python
for i in range(5, 0, -1):  
    print((str(i) + " ") * (6 - i))  
```
**Output:**  
```
5  
4 4  
3 3 3  
2 2 2 2  
1 1 1 1 1  
```

### **3️⃣ Print Prime Numbers in a Range**  
```python
start = int(input("Start: "))  
end = int(input("End: "))  

print("Prime Numbers:")  
for num in range(start, end + 1):  
    if num > 1:  
        for i in range(2, num):  
            if num % i == 0:  
                break  
        else:  
            print(num, end=" ")  
```

## **📌 Week 5: Lists, Tuples & Dictionaries**  

### **1️⃣ List vs Tuple**  
```python
# List (Mutable - Can Change)  
fruits = ["Apple", "Banana", "Cherry"]  
fruits.append("Mango")  

# Tuple (Immutable - Cannot Change)  
colors = ("Red", "Green", "Blue")  
# colors[0] = "Yellow"  ❌ Error!  
```

### **2️⃣ Find Common Elements in Two Lists**  
```python
list1 = [1, 2, 3, 4]  
list2 = [3, 4, 5, 6]  
common = [x for x in list1 if x in list2]  
print("Common Elements:", common)  
```

### **3️⃣ Dictionary Example**  
```python
student = {"name": "Alice", "age": 20, "grade": "A"}  
print(student["name"])  # Output: Alice  
student["age"] = 21    # Update value  
print(student)  
```

## **📌 Week 6: Functions & Modules**  

### **1️⃣ Check if a List is Sorted**  
```python
def is_sorted(lst):  
    for i in range(len(lst) - 1):  
        if lst[i] > lst[i+1]:  
            return False  
    return True  

numbers = [1, 3, 5, 7]  
print(is_sorted(numbers))  # True  
```

### **2️⃣ GCD of Two Numbers**  
```python
def gcd(a, b):  
    while b:  
        a, b = b, a % b  
    return a  

print("GCD:", gcd(48, 18))  # Output: 6  
```
