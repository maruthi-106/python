**1.i) Compound Interest**  
```python
P = float(input("Principal: "))
r = float(input("Rate (%): "))
t = float(input("Time (years): "))
amount = P * (1 + r/100) ** t
print("Final Amount:", amount)
```

**1.ii) Distance Between Points**  
```python
import math
x1, y1 = map(float, input("Point 1 (x1 y1): ").split())
x2, y2 = map(float, input("Point 2 (x2 y2): ").split())
distance = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)
print("Distance:", distance)
```

### **Week 4: Loop Statements**  
**1) Fibonacci Sequence**  
```python
n = int(input("Terms: "))
a, b = 0, 1
for _ in range(n):
    print(a, end=" ")
    a, b = b, a + b
```

**2) Number Triangle**  
```python
for i in range(5, 0, -1):
    print((str(i) + " ") * (6 - i))
```

**3) Prime Numbers in Interval**  
```python
start, end = map(int, input("Range (start end): ").split())
for num in range(start, end + 1):
    if num > 1:
        for i in range(2, num):
            if num % i == 0:
                break
        else:
            print(num, end=" ")
```

### **Week 6: Functions & Modules**  
**1) Check Sorted List**  
```python
def is_sorted(lst):
    return all(lst[i] <= lst[i + 1] for i in range(len(lst) - 1))
print(is_sorted([1, 2, 3]))  # True
```

**2) GCD Function**  
```python
def gcd(a, b):
    while b:
        a, b = b, a % b
    return a
print("GCD:", gcd(48, 18))  # Output: 6
```

**3) Geometry Module Example**

**3) Custom Title Case**  
```python
def custom_title(sentence):
    return " ".join(word[0].upper() + word[1:].lower() for word in sentence.split())
print(custom_title("hello world"))  # Output: Hello World
```

### **Week 8: Classes & Objects**  
**1) Complex Numbers**  
```python
class Complex:
    def __init__(self, real, imag):
        self.real = real
        self.imag = imag
    def __add__(self, other):
        return Complex(self.real + other.real, self.imag + other.imag)

c1 = Complex(1, 2)
c2 = Complex(3, 4)
result = c1 + c2
print(f"{result.real} + {result.imag}i")  # Output: 4 + 6i
```

### **Week 9: Inheritance**  
**1) Types of Inheritance**  
```python
class Parent:
    def greet(self):
        print("Hello from Parent!")

class Child(Parent):  # Single Inheritance
    def greet(self):
        super().greet()
        print("Hello from Child!")

obj = Child()
obj.greet()
```

### **Week 10: File Handling**  
**1) Merge Two Files**  
```python
with open("file1.txt") as f1, open("file2.txt") as f2, open("merged.txt", "w") as out:
    out.write(f1.read() + f2.read())
```

**2) Most Frequent Word**  
```python
from collections import Counter
words = open("file.txt").read().split()
print("Most frequent:", Counter(words).most_common(1)[0][0])
```

**3) Count Words, Vowels, Spaces**  
```python
text = open("file1.txt").read()
print("Words:", len(text.split()))
print("Vowels:", sum(1 for c in text.lower() if c in "aeiou"))
print("Spaces:", text.count(" "))
```
