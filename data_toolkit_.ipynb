```python
#How do you calculate the correlation matrix using Seaborn and visualize it with a heatmap
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=sns.load_dataset('titanic')
num_df= df[['age','fare','pclass','sibsp','parch']]
cor_matrix = num_df.corr()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(cor_matrix,annot=True,cmap='coolwarm',fmt=".2f")
plt.title("correlation matrix heatmap")
plt.show()
```

```python
# Create a line plot with multiple lines using MatplotlibA
import matplotlib.pyplot as plt

# Sample data
x = [1, 2, 3, 4, 5]
y1 = [1, 4, 9, 16, 25]     # Line 1
y2 = [1, 3, 6, 10, 15]     # Line 2
y3 = [2, 3, 5, 7, 11]      # Line 3

# Create the plot
plt.plot(x, y1, label='Squares', color='blue', marker='o')
plt.plot(x, y2, label='Triangular Numbers', color='green', linestyle='--', marker='x')
plt.plot(x, y3, label='Primes', color='red', linestyle='-.', marker='s')

# Add labels and title
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.title('Multiple Lines on One Plot')

# Add legend
plt.legend()

# Show the plot
plt.grid(True)
plt.show()
```

```python
#Create a histogram using Seaborn to visualize a distributionA
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

data = {
    'Marks': [45, 67, 89, 76, 83, 95, 62, 70, 88, 91, 58, 72, 65, 80]
}
df = pd.DataFrame(data)

sns.histplot(df['Marks'],bins=6,kde=True,color='skyblue')
plt.xlabel("marks")
plt.ylabel("frequency")
plt.title("Distribution of marks")
plt.grid(True)
plt.show()
```

```python
#<<A Perform matrix multiplication using NumPyA
import numpy as np

# Define Matrix A (2x3)
A = np.array([[1, 2, 3],
              [4, 5, 6]])

# Define Matrix B (3x2)
B = np.array([[7, 8],
              [9, 10],
              [11, 12]])

# Perform matrix multiplication
result = np.dot(A, B)  # or use: result = A @ B

# Display the result
print("Matrix A:\n", A)
print("Matrix B:\n", B)
print("Matrix A x B:\n", result)
```

```python
#<>A Create a 3D scatter plot using Plotly.
import plotly.graph_objects as go

x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]
z = [5, 15, 25, 35, 45]

fig = go.Figure(data=[go.Scatter3d(
    x=x,
    y=y,
    z=z,
    mode='markers',
    marker=dict(
        size = 8,
        color = z,
        colorscale = 'Viridis',
        opacity = 0.8
    )
)])


fig.update_layout(
    title = '3D Scatter Plot Example',
    scene = dict(
        xaxis_title = 'X Axis',
        yaxis_title = 'Y Axis',
        zaxis_title = 'Z Axis'
    )
)


fig.show()
```
