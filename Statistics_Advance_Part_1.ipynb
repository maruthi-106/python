```python
#3. Write a Python function to calculate the probability distribution function (PDF) of a Bernoulli distribution
def ber_pdf(x,p):
  if x not in [0,1]:
    raise ValueError("x must be 0 or 1 for bernoulli distribution.")
  if not (0 <= p <= 1):
    raise ValueError("p must be between 0 and 1.")
  return p if x==1 else 1-p
for x in [0, 1]:
    print(f"P(X={x}) =", ber_pdf(x, 0.3))
```

```python
#6. Write a Python program to calculate and plot the cumulative distribution function (CDF) of a discrete uniform distribution
import numpy as np
import matplotlib.pyplot as plt

a=1 #min value
b=6 #max value(like a die)

x = np.arange(a,b+1)

pdf = np.ones_like(x)/len(x)
cdf = np.cumsum(pdf)

plt.step(x,cdf,where ='mid',label='CDF',color='blue')
plt.scatter(x,cdf,color='red')
plt.title('CDF of Discrete Uniform Distribution')
plt.xlabel('x')
plt.ylabel('Cumulative Probability')
plt.xticks(x)
plt.grid(True,linestyle='--',alpha=0.7)
plt.legend()
plt.show()
```

```python
#9.  Write a Python function to calculate Z-scores from a dataset and plot them

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

def plot_z_scores(data):

    data = np.array(data)

    mean = np.mean(data)
    std = np.std(data)

    z_scores = (data - mean) / std


    plt.hist(z_scores, bins=30, density=True, alpha=0.6, color='skyblue', edgecolor='black', label='Z-scores')

    x = np.linspace(-4, 4, 500)
    plt.plot(x, norm.pdf(x), 'r-', label='Standard Normal PDF')

    plt.title('Z-Score Distribution')
    plt.xlabel('Z-score')
    plt.ylabel('Density')
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.legend()
    plt.show()

sample_data = np.random.normal(loc=50, scale=10, size=1000)
plot_z_scores(sample_data)

```

```python
# Write a Python program to calculate the Z-score for a given data point and compare it to a standard normal distribution

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

def cal_z_score(x,mu,sigma):
  return (x-mu)/sigma

def plot_std_nor_with_z(z_score):


  x = np.linspace(-4,4,1000)
  pdf = norm.pdf(x)

  plt.figure(figsize=(10,5))
  plt.plot(x,pdf,color ='navy',label='standerd normal pdf')

  x_fill = np.linspace(-4,z_score,1000)
  plt.fill_between(x_fill,norm.pdf(x_fill),linestyle = '--',label = f'z-score = {z_score}')
  plt.title('standerd normal distribution with z-score highlighted')
  plt.xlabel('Z')
  plt.ylabel('probability Density')
  plt.legend()
  plt.grid(True)
  plt.show()

x = 115
mu = 100
sigma = 15

z = cal_z_score(x,mu,sigma)
print(f'Z-score for x = {x} is {z:.2f}')


plot_std_nor_with_z(z)
```

```python
# Implement hypothesis testing using Z-statistics for a sample dataset
import numpy as np
from scipy.stats import norm

sample = np.array([102,100,98,101,99,97,103,100,98,102])
sample_mean = np.mean(sample)
sample_size = len(sample)


mu_0 = 100
sigma = 2
alpha = 0.05

z = (sample_mean - mu_0)/(sigma/np.sqrt(sample_size))

p_value = 2*(1-norm.cdf(abs(z)))

print(f'Sample mean = {sample_mean:.2f}')
print(f'Z-statistic = {z:.4f}')
print(f'p-value = {p_value:.4f}')

if p_value < alpha:
  print('reject the null hypothesis: significant difference')
else:
  print('fail to reject the null hypothesis: no significant difference')
```

```python
# Create a confidence interval for a dataset using Python and interpret the result
import numpy as np
import scipy.stats as stats

np.random.seed(42)
data = np.random.exponential(scale = 2.0,size = 100)


sample_mean = np.mean(data)
sample_std = np.std(data,ddof=1)

n = len(data)

standard_error = sample_std / np.sqrt(n)

confidence_level  = 0.95
z_score = stats.norm.ppf((1+confidence_level)/2)# ≈ 1.96

margin_of_error = z_score * standard_error
ci_lower = sample_mean - margin_of_error
ci_upper = sample_mean + margin_of_error

print(f"sample mean: {sample_mean:.3f}")
print(f"95% confidence interval: ({ci_lower:.3f},{ci_upper:.3f})")
```

```python
# Use Python to calculate and interpret the cumulative distribution function (CDF) of a Poisson distribution
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

lambda_param = 4
x = np.arange(0,15)

cdf_values = poisson.cdf(x,mu=lambda_param)

k = 6
prob = poisson.cdf(k,mu = lambda_param)
print(f"P(X ≤ {k}) = {prob:.4f}")

plt.figure(figsize=(8, 5))
plt.step(x, cdf_values, where='post', color='green', label=f'Poisson CDF (λ={lambda_param})')
plt.scatter(x,cdf_values,color='red')
plt.title('Cumulative Distribution Function of Poisson Distribution')
plt.xlabel('Number of Events (k)')
plt.ylabel('P(X ≤ k)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()
```

```python
# Implement Z-score normalization on a dataset and visualize its transformation
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
data = np.random.normal(loc=50, scale=10, size=1000)

mean=np.mean(data)
std=np.std(data)

z_scores = (data - mean)/std

plt.figure(figsize=(12, 5))

# Original data
plt.subplot(1, 2, 1)
plt.hist(data, bins=30, color='skyblue', edgecolor='black')
plt.title('Original Data')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.grid(True, linestyle='--', alpha=0.6)

# Normalized data
plt.subplot(1, 2, 2)
plt.hist(z_scores, bins=30, color='salmon', edgecolor='black')
plt.title('Z-score Normalized Data')
plt.xlabel('Z-score')
plt.ylabel('Frequency')
plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

```python
# Write a Python function to calculate the skewness and kurtosis of a dataset generated from a normal distribution
import numpy as np
from scipy.stats import skew, kurtosis

def analyze_normal_distribution(mu=0, sigma=1, size=1000):
    
    data = np.random.normal(loc=mu, scale=sigma, size=size)
    
    skewness = skew(data)
    excess_kurtosis = kurtosis(data)  # By default, this is excess kurtosis (kurtosis - 3)
    
    print(f"Skewness: {skewness:.4f}")
    print(f"Excess Kurtosis: {excess_kurtosis:.4f}")
    
    return skewness, excess_kurtosis


analyze_normal_distribution(mu=50, sigma=10, size=1000)
```
